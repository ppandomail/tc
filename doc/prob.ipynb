# Problemas, lenguajes y decidibilidad
---

* ¿Qué cosas las MT **NO pueden hacer**?
* Los problemas pueden clasificarse según exista o no un mecanismo efectivo para resolverlos
* Un problema se describe con un lenguaje => cuanto más formal el lenguaje, más precisa la formulación del problema

## Problemas de decisión

* En teoría de la computación, muchos problemas pueden expresarse como preguntas cuya respuesta es sí/no
* Todo problema de decisión puede asociarse a un lenguaje:

$$L=\{w\in\Sigma^* : w \text{ es una instancia cuya respuesta es sí}\}$$

* Por tanto:
    * Resolver un problema de decisión equivale a decidir la pertenencia de una cadena a un lenguaje

* La Teoría de Clasificación de Problemas se basa en el estudio de los Problemas de Decisión
* Todo problema de salida general con un cierto grado de dificultad, se puede transformar en un problema de decisión del mismo grado de dificultad (Reducción de problemas)
* Ejemplos
  * Determinar un camino entre dos ciudades => Problema salida general
  * ¿Hay un camino entre dos ciudades? => Problema de decisión

### Ejemplo de problema de decisión:

$$\text{"¿Este número entero n es par?"}$$
* **Lenguaje**: la colección de todas las respuestas "SÍ" codificadas como texto
* Si codificamos en binario, el lenguaje de los números pares sería:

$$L = \{0, 10, 100, 110, \dots\}$$ 
$$\text{(cadenas que terminan en 0)}$$

* **Decidible**: significa que es posible escribir un programa que siempre termine y dé la respuesta correcta (SÍ o NO) para cualquier entrada
* **Indecidible**: un problema para el cual es matemáticamente imposible escribir un programa general que siempre acierte y nunca se cuelgue en un bucle infinito

## 1. Problemas decidibles o solubles

* Son aquellos para los que existe un algoritmo capaz de determinar la respuesta y detenerse para cualquier entrada
* Existe un ALGORITMO TOTAL que para TODA INSTANCIA del problema devuelve la RESPUESTA CORRECTA (responder sí o no a cada caso)
* Existe una MT que **siempre para** al resolver el problema
* Ejemplos:
    * Dado un número natural n, decidir si n es par o no
    * Dados dos AF, decidir si son equivalentes
    * Dada una cadena arbitraria w y un LR L, determinar si w pertenece o no a L
    * Dada una GIC G, ¿L(G) es vacío?. Hay un número infinito de casos, uno por cada GIC. Además, para cada caso, se puede determinar una respuesta afirmativa o negativa
* Se corresponden con los lenguajes recursivos (LRec)

$$\text{Problema decidible} <=> \text{Lenguaje recursivo}$$

* Un lenguaje $L$ es recursivo si existe una Máquina de Turing $M$ tal que:

$$
M(w)=
\begin{cases}
\text{acepta} & w \in L \\
\text{rechaza} & w \notin L
\end{cases}
$$

$$ \text{y, fundamentalmente, la máquina siempre se detiene}$$

```
Entrada w
       ↓
Máquina de Turing
       ↓
¿w pertenece a L?
 ↙             ↘
SÍ             NO
↓              ↓
acepta      rechaza
  \            /
  siempre termina
```
  
* Dentro de los lenguajes estudiados anteriormente encontramos:
    * Lenguajes regulares (LR)
    * Lenguajes independientes del contexto (LIC)
    * Lenguajes dependientes del contexto (LDC)
* Estos lenguajes son recursivos, por lo que sus problemas de pertenencia son decidibles

```
LRE
┌─────────────────────────────────┐
│                                 │
│ REC (decidibles)                │
│ ┌─────────────────────────┐     │
│ │ LDC                     │     │
│ │ ┌───────────────────┐   │     │
│ │ │ LIC               │   │     │
│ │ │ ┌─────────────┐   │   │     │
│ │ │ │ LR          │   │   │     │
│ │ │ └─────────────┘   │   │     │
│ │ └───────────────────┘   │     │
│ └─────────────────────────┘     │
└─────────────────────────────────┘
```

### Ejemplo de problema decidible

* **Pregunta**: dada una palabra $w$, ¿es un palíndromo (se lee igual al revés)?
* **Algoritmo**:
    1. Comparar la primera letra con la última
    2. Si coinciden, avanzar hacia el centro
    3. Si todas coinciden, responder **SÍ**
    4. Si alguna falla, responder **NO**
* **Conclusión**: como la palabra tiene longitud finita, el algoritmo siempre termina en tiempo finito. El problema es **decidible**

## 2. Problemas indecidibles o insolubles

* Son problemas para los cuales no existe un algoritmo que siempre termine y proporcione una respuesta correcta para todas las entradas
* No existe un PROCEDIMIENTO EFECTIVO para determinar si la propiedad es verdadera
* **No existe una MT para resolver el problema**
* Ejemplos:
    * Decidir si una cadena pertenece o no pertenece a un LRE
    * Dada una GIC G, ¿G es ambigua?
    * Problema de Correspondencia Post
    * Dada una MT M, ¿M se detendrá comenzando en el estado inicial con w en la cinta? (Problema del Halting)
* Aquí aparecen los lenguajes recursivamente enumerables (LRE) que no son recursivos
* Comportamiento de la MT
$\text{Si } M \text{ reconoce } L:$

$$w \in L\Rightarrow M(w)\text{ acepta y se detiene}$$

$\text{pero:}$

$$w \notin L\Rightarrow \begin{cases} M(w)\text{ rechaza},\\ \text{o}\\ M(w)\text{ puede ejecutarse indefinidamente}. \end{cases}$$
$$\text{Esta última posibilidad es la diferencia fundamental respecto de los lenguajes recursivos}$$

### Lenguajes recursivamente enumerables (LRE)

* Un lenguaje $L$ es recursivamente enumerable si:
    * Existe una Máquina de Turing (MT) capaz de enumerar las palabras de $L$
    * Equivalentemente, existe una MT que reconoce las cadenas pertenecientes a $L$:
        * Si $w \in L$, la máquina se detiene y acepta
        * Si $w \notin L$, la máquina no necesariamente se detiene
    * A estos lenguajes se les pueden asociar procedimientos, pero no necesariamente algoritmos de decisión
* Por tanto: $LR \subsetneq LRE$
* Es decir, **todo lenguaje recursivo es recursivamente enumerable, pero no todo lenguaje recursivamente enumerable es recursivo**

### Lenguajes no recursivamente enumerables (LNRE)

* Existen lenguajes que no son recursivamente enumerables
* En estos casos, no existe una Máquina de Turing que pueda reconocer todas y únicamente las cadenas pertenecientes al lenguaje
* Por tanto, tenemos tres situaciones conceptualmente diferentes:
  
```
Lenguajes
│
├── Recursivos
│ └── Decidibles → Algoritmos
│
├── LRE no recursivos
│ └── Indecidibles → Procedimientos de reconocimiento
│
└── No recursivamente enumerables (LNRE)
```

## Relación con los lenguajes de programación

* Si un lenguaje es LRE pero no recursivo, puede existir un procedimiento para reconocer o enumerar sus cadenas, pero no existe un algoritmo que decida la pertenencia para cualquier cadena
* Esto resulta problemático para un lenguaje de programación, ya que necesitamos poder determinar si un programa pertenece al lenguaje, es decir, si es sintácticamente válido
* Idea central:

$$\text{Decidible} => \text{Recursivo} => \text{Recursivamente enumerable}$$

$$\text{Pero las implicaciones inversas no se cumplen en general}$$

* Un compilador debe poder determinar, entre otras cosas, si una cadena cumple determinadas reglas sintácticas
* Para la sintaxis de un lenguaje de programación necesitamos procedimientos efectivos que terminen
* Por ello, los lenguajes utilizados para describir la sintaxis suelen pertenecer a clases mucho más restringidas, como los lenguajes independientes del contexto, y por tanto son decidibles
* No confundir:
    * decidir si un texto es un programa sintácticamente válido
    * decidir qué hará ese programa cuando se ejecute
* La primera cuestión puede ser decidible mientras que determinadas propiedades del comportamiento del programa pueden ser indecidibles

## Clasificación final

```
PROBLEMAS
│
├── DECIDIBLES / SOLUBLES
│      │
│      └── Lenguajes recursivos
│             │
│             ├── siempre aceptan o rechazan
│             └── siempre terminan
│
└── INDECIDIBLES / INSOLUBLES
        │
        ├── LRE no recursivos
        │      ├── reconocibles
        │      ├── semidecidibles
        │      └── pueden no terminar
        │
        └── LNRE
               └── ni siquiera son reconocibles por una MT
```

### Ejemplo de problema indecidible: Problema de Correspondencia Post

* Dadas dos listas A y B de cadenas sobre un alfabeto, determinar si existe una concatenación de cadenas A y B de la forma:
    * xi1 xi2 ... xim = wi1 wi2 ... wim donde xij ∈ A y wij ∈ B (si elijo el indice i para x elijo el mismo indice i para w)

* **Ejemplo 1**:

| Lista A | Lista B |
| -- | -- |
| Xi | Wi |
| 1 | 111 |
| 10111 | 10 |
| 10 | 0|

* x2 x1 x1 x3 = w2 w1 w1 w3
  * x2 x1 x1 x3 = 10111 1 1 10
  * w2 w1 w1 w3 = 10 111 111 0
* Para esta entrada, si tomamos los indices 2, 1, 1, 3 se obtiene una solución del problema

* **Ejemplo 2**:

| Lista A | Lista B |
| -- | -- |
| Xi | Wi |
| 10 | 101 |
| 011 | 11 |
| 101 | 011 |

* x1 x3 x1 <> w1 w3 w1
  * x1 x3 x1 = 10 101 10
  * w1 w3 w1 = 101 011 101

* x1 x3 x3 x3 <> w1 w3 w3 w3
  * x1 x3 x3 x3 = 10 101 101 101
  * w1 w3 w3 w3 = 101 011 011 011

* Se debe comenzar con la primera cadena de ambas listas
* Después la única elección posible es la tercera cadena, pero siempre queda un "uno de mas" en la cadena armada a partir de la lista B
* Se puede seguir infinitamente buscando si existe solución

* Para algunas instancias de las listas A y B se puede obtener una respuesta
* Para otras instancias de las listas A y B algunas veces no se puede dar respuesta
* Problema de Correspondencia Post es indecidible

### Ejemplo de problema indecidible: Problema de la parada

* Sea M una MT arbitraria con un alfabeto de entrada Σ. Sea w ∈ Σ*.  ¿Parará M con la palabra w como cadena de entrada?, es decir, ¿se puede determinar si M terminará en un número finito de pasos cuando es ejecutada usando w como palabra de entrada?
* Alan Turing, en su famoso artículo “On Computable Numbers, with an Application to the Entscheidungsproblem” (1936), demostró que el problema de la parada de la MT es indecidible, en el sentido de que ninguna MT lo puede resolver
* Al ejecutar un programa, este puede terminar  después de un número finito de pasos o puede no terminar nunca (queda “trabado” o que entra en un bucle infinito)
* ¿Existe  un programa P, tal que, dado un programa cualquiera Q y unos datos de entrada x, muestre como salida:
  * 1 si Q termina en número finito de pasos o
  * 0 si Q entra en un bucle infinito

```sh
Algoritmo Detención MT
DatoEntrada: MT T, w
DatoSalida: Estado
Comienzo
   ...
   Devolver Estado con valor "se detiene" o "cicla"
Fin
```

* **Turing ha demostrado que este algoritmo NO EXISTE**
* Conocer si existe el programa P es, en términos resumidos, el problema de la parada
* No existe una manera automática computable de saber si todos los programas del mundo terminan
* No se niega que exista la prueba para programas concretos
* De hecho, la construcción de pruebas para programas concretos es un paso obligatorio para demostrar su correctitud

* **Pregunta**: ¿Se puede crear un antivirus o verificador perfecto que lea el código fuente de cualquier programa y determine si ese programa terminará o se quedará colgado para siempre en un ```while (true)```?

* **Respuesta de Turing (1936)**: NO. No importa qué tan potente sea la computadora o qué tan inteligente sea el algoritmo, siempre existirá algún programa especialmente diseñado para engañar al verificador (creando una paradoja lógica del tipo "si dices que paro, entro en un bucle; si dices que entro en bucle, paro")
