# TP Máquinas Secuenciales
---

### Detector de la secuencia 01

* Diseñar una máquina que emita:
    * Salida 1 cuando detecte la secuencia 01
    * Salida 0 en cualquier otro caso

### Paridad par de unos 

* Diseñar una máquina que indique si después de leer un símbolo la cantidad de unos es par
* La salida vale 1 cuando el número de unos leídos es par

### Máquina expendedora simple

* Una máquina acepta monedas de:
    * 5 pesos
    * 10 pesos
* El producto cuesta 15 pesos
* Emitir salida 1 cuando se alcance o supere el importe

### Detector de la secuencia 101

* Construir una máquina que produzca salida 1 cada vez que aparezca la secuencia "101"
* Ejemplo:
    * Entrada:  110101
    * Salida: 000101 (porque la secuencia 101 aparece dos veces) 

### Detector de dos ceros consecutivos

* Emitir salida 1 cada vez que aparezcan dos o más ceros consecutivos
* Ejemplo:
    * Entrada: 1000010
    * Salida:  0001110

### Detector de patrón binario palindrómico de longitud 3

* Detectar cada aparición de:
    * 000
    * 010
    * 101
    * 111
* Emitir salida 1 cuando aparezcan
* Pista: se necesita recordar los dos últimos bits
