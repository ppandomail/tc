# Máquina de Turing Universal (MTU)
---

* Es un modelo teórico de cómputo que puede **simular cualquier otra MT**
* Recibe el nombre de universal, porque es capaz de simular el comportamiento de cualquier MT
* Fue propuesta por **Alan Turing en 1936** como parte de su trabajo sobre la decidibilidad y la computabilidad
* Alan Turing introdujo una generalización de su máquina, a la que se denominó MTU o máquina-u
* También llamada Máquina de Propósito General, pues no fue construida para un problema específico
* En cierto sentido: La MTU es una máquina que puede ejecutar otras máquinas como si fueran programas
* Esta idea también es fundamental para comprender conceptos como programabilidad, intérpretes, computabilidad y el problema de la parada

## Idea central

* La MTU recibe como entrada:
  1. La **descripción codificada** de otra MT $M$
  2. Una **cadena de entrada** $w$ para esa máquina $M$
* La MTU procesa la descripción y simula paso a paso el comportamiento de $M$ sobre $w$; es decir, el mismo resultado que produciría la MT sobre su cinta
* Se convierte de esta forma, en un dispositivo análogo a una computadora, capaz de realizar cualquier cómputo

![MTU](img/mtu.png)

## Formalización

* Sea $U$ la máquina de Turing universal
* Entrada: $\langle M,w \rangle$, donde:
  * $\langle M\rangle$ es la codificación de la máquina $M$
  * $w$ es la cadena de entrada
* Comportamiento:

$$U(\langle M, w\rangle) = M(w)$$

* Es decir, $U$ acepta si $M$ acepta $w$, y rechaza si $M$ rechaza $w$

## Ejemplo

* MT M suma dos números. Se puede codificar sus estados, símbolos y transiciones:
	* $\langle M\rangle$ = "descripción de M"
* MTU recibe:
	* $\langle M\rangle$, 3+5
* y simula a M, produciendo:
	* 8
* Lo importante es que la máquina universal no necesita estar diseñada específicamente para sumar
* Si en cambio, se le dá la descripción de una máquina que multiplica, ordena números o reconoce un determinado lenguaje, puede simularla también

## Propiedades fundamentales

* **Simulación universal:** cualquier máquina de Turing puede ser simulada por la MTU
* **Base de la computación moderna:** la idea de una máquina universal inspiró el concepto de **computadora programable**
* **Codificación:** se requiere un método para representar estados, símbolos y transiciones de $M$ como datos que $U$ pueda leer
* **Relación con el problema de la parada:** la existencia de $U$ muestra que podemos simular cualquier máquina, pero no garantiza decidir si se detendrá

## Un poco de historia

* En 1945, el físico e ingeniero John Von Neumann redacta el primer borrador de lo que luego sería la EDVAC **Idea: construir una computadora electrónica basada en la Máquina Universal de Turing (MUT), con la noción de "programa almacenado"**
* En 1948, Frederic C. Williams, Tom Kilburn and Geoff Tootill construyen la **Manchester Small-Scale Experimental Machine (SSEM)** en la Univ. de Manchester (Inglaterra), apodada **"Baby Machine"**. Es la primer computadora que tiene un programa almacenado. Podía guardar 32 instrucciones. El 21 de junio de 1948 se ejecuta por primera vez un programa almacenado
* La máquina está reconstruida en el Museum of Science & Engineering de Manchester, y puede verse en funcionamiento todos los martes
* El primer programa almacenado tenía 17 instrucciones
* Escrito por Tom Kilburn (1921 - 2001), fue diseñado para encontrar el factor más grande de 2^18 probando todos los números desde 2^18 – 1 hacia abajo
* Ejecutarlo llevó 3,5 millones de operaciones, y tardó 52 minutos

## Importancia histórica

* Introduce el concepto de **programa almacenado**: los algoritmos pueden representarse como datos.
* Es el fundamento teórico de la **arquitectura de von Neumann**.
* Marca el inicio de la teoría de la **computabilidad** y de los **lenguajes formales**.

## Ejemplo de codificación

### Contexto

Para que la **MTU** pueda simular otra MT $M$, se necesita **codificar**:
* El **alfabeto de entrada** y el **alfabeto de cinta**
* Los **estados** de la MT $M$
* Las **transiciones** (función de transición)

### 1. Definición de la MT $M$

$M = (\{0, 1\}, 0, \{p, q, r\}, p, \delta, \{\})$

| Q  | 0   | 1   |
| -- | --  | --  |
| p  | q0R | p1L |
| q  | p1L | r0R |
| r  | q1R | p0L |

### 2. Codificación de elementos

Se pueden usar números binarios para codificar:
* Símbolos:  
  * `0 = 0`, `1 = 1`
* Estados:  
  * `p = 00`, `q = 01`, `r = 10`
* Movimientos:  
  * `L = 1`, `R = 0`

### 3. Codificación de transiciones

Cada transición se representa como una **tupla codificada**:

| Q   | 0    | 1    |
| --  | --   | --   |
| 00  | 0100 | 0011 |
| 01  | 0011 | 1000 |
| 10  | 0110 | 0001 |

* $\delta(p, 0) = (q, 0, R)$  
  * Codificación: `(00, 0, 01, 0, 0)`

### 4. Entrada para la MTU

$\langle M, w \rangle$


donde:
- $\langle M \rangle$ es la lista de transiciones codificadas
- $w$ es la cadena de entrada, por ejemplo `101101`

Ejemplo de entrada codificada:

`101*01$001#0000100#0010011#0100011#0111000#1000110#10100001`

* La cinta de una MTU es infinita en ambos sentidos, y se divide en dos partes, separadas por símbolo identificativo (por ejemplo, $\$$)
* A la derecha de $\$$ se codifica la MT $M$
* A la izquierda de $\$$ se codifica su cinta $w$
* El resultado de la ejecución de la MTU consistirá en dejar la parte derecha de la cinta (la que describe la MT $M$) intacta, y, en la izquierda, la misma información que habría dejado en su cinta la MT $M$
* La codificación de la MT $M$ transcribe la cinta tal cual, exceptuando la posición que ocupa la cabeza lectora grabadora, en la cual se coloca otro símbolo identificativo de cabeza lectora grabadora (por ejemplo,  $*$)
* El bit que había en dicha posición, se transcribe justo a la derecha del símbolo $\$$
* La parte derecha de la cinta se divide en tantos registros como entradas tenga la tabla de transiciones de la MT $M$
* Cada registro está separado por un símbolo especial (por ejemplo, $\#$) y contiene, en binario, la codificación de cada transición
* Entre el símbolo $\$$ y el primer registro, denotado por $\#$ está el estado en que está la máquina y el símbolo que lee en ese momento la cabeza lectora grabadora

### 5. Interpretación

**La MTU leerá esta codificación y simulará paso a paso la ejecución de $M$ sobre la cadena `101101`**

### 6. Funcionamiento

1. Localizar en alguno de los registros la cadena que aparece entre los símbolos $\$$ y $\#$ (llamarlo registro 0)
1. Leer y memorizar uno a uno los símbolos que restan del registro encontrado, excepto el último (que se corresponde con el movimiento a realizar)
1. Para c/u de los símbolos memorizados con anterioridad, colocarlos en la parte correspondiente del registro 0
1. El penúltimo bit del registro localizado (que se corresponde con el bit a escribir), se escribe en el lugar que ocupa el $*$
1. Se localiza y memoriza el último bit del registro encontrado anteriormente (que se corresponde con el movimiento), para desplazar la cabeza lectora/grabadora correctamente y colocar el $*$ en el lugar adecuado
1. El contenido que había en la posición donde se escribe el $*$ ha de ser memorizado y escrito en la última posición del registro 0, para, de esta forma, recoger lo que lee la cabeza lectora y poder preparar todo el sistema para una nueva lectura
