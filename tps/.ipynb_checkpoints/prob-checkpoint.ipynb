## De problema a lenguaje

**Define formalmente como lenguaje el problema: "¿Contiene una cadena sobre el alfabeto $\{a, b\}$ una cantidad par de letras 'a'?"**

El lenguaje $L$ es el conjunto de cadenas válidas:
$$L = \{ w \in \{a, b\}^* \mid \text{el número de apariciones de 'a' en } w \text{ es par} \}$$
* Las cadenas $b$, $aa$, $aba$, $baab$ pertenecen a $L$
* Las cadenas $a$, $abb$, $aaa$ no pertenecen a $L$

## ¿Decidible o Indecidible?

**Clasifica los siguientes 3 problemas justificando brevemente:**
1. **Dado un número entero positivo $x$, ¿es un número primo?**
2. **Dado un código en Python y una entrada $y$, ¿el programa imprime la palabra "Hola" al ejecutarse?**
3. **Dado un texto $T$ y una palabra $p$, ¿aparece $p$ dentro de $T$?**

1. **Decidible**: Basta con probar divisiones desde $2$ hasta $\lfloor\sqrt{x}\rfloor$. El proceso siempre termina

2. **Indecidible**: Es una variante directa del problema de la parada. Para saber con certeza si llegará a esa línea de código en cualquier caso posible, tendríamos que predecir la ejecución completa sin riesgo de quedar atrapados en un bucle infinito

3. **Decidible**: Es una búsqueda de subcadenas estándar (como la función Ctrl + F). La longitud del texto es finita, por lo que el algoritmo siempre concluye

## Demostrar que es decidible

**Problema**: Verificar si una cadena de paréntesis está bien balanceada

$$L = \{ w \in \{'(', ')'\}^* \mid w \text{ tiene paréntesis balanceados} \}$$

* Una definición inductiva precisa es:
    1. La cadena vacía $\lambda \in L$
    2. Si $u \in L$ y $v \in L$, entonces $uv \in L$ (concatenación)
    3. Si $u \in L$, entonces $(u) \in L$ (anidamiento)
    4. Nada más pertenece a $L$
* Pertenecen a $L$: (), (()), ()(), ((())())
* No pertenecen a $L$: )(, ((), ()), )(()
* Como la cadena de entrada tiene una longitud finita $n$, el algoritmo hace exactamente una operación de pila por cada símbolo leído. Termina siempre en exactamente $n$ pasos, respondiendo con certeza SÍ o NO

## Demostrar que es indecidible

**Problema**: Determinar si un programa llega alguna vez a imprimir "Hola Mundo"

$$L_{\text{hola}} = \{ \langle P, x \rangle \mid \text{el programa } P \text{ con la entrada } x \text{ imprime } \text{"Hola Mundo"} \}$$

Para demostrar que un problema nuevo $B$ es imposible de resolver, mostramos que si tuviéramos una solución para $B$, podríamos resolver un problema $A$ que ya sabemos que es imposible (como el Problema de la Parada, $HALT$)

$$\text{Si } B \text{ fuera decidible} \implies HALT \text{ sería decidible} \quad (\text{¡Contradicción!})$$

* Solución Paso a Paso
1. Partir del supuesto contrario (Absurdo)
    * Supongamos que existe una función o verificador perfecto llamado ```DecisorHola(P, x)``` que:
        * Responde SÍ si el programa ```P``` con entrada ```x``` imprime ```"Hola Mundo"```
        * Responde NO si nunca lo imprime (ya sea porque termina sin imprimirlo o porque entra en un bucle infinito antes de llegar a esa línea)
        * ```DecisorHola``` siempre termina en tiempo finito
2. Construir la trampa para resolver el Problema de la Parada
    * Queremos resolver el problema clásico de la parada: saber si un programa arbitrario ```M``` con entrada ```w``` se detiene o se cuelga para siempre
    * Diseñamos un transformador que toma el código de ```M``` y los datos ```w```, y genera el código fuente de un nuevo programa ```M'``` de la siguiente manera:

````python
def programa_M_prima(entrada_cualquiera):
    # 1. Eliminamos o silenciamos cualquier print("Hola Mundo") que M tuviera originalmente.
    # 2. Ejecutamos M con la entrada w:
    ejecutar M(w)
    
    # 3. Solo si M(w) termina con éxito, llegamos a esta línea:
    print("Hola Mundo")
````

3. Analizar el comportamiento de ```M'```
    * Observemos qué hace ```M'```:
        * Caso 1: Si ```M(w)``` se detiene, la línea 2 termina y el flujo pasa a la línea 3 $\implies$ ```M'``` imprime ```"Hola Mundo"```
        * Caso 2: Si ```M(w)``` entra en un bucle infinito, nunca pasa de la línea 2 y jamás llega a la línea 3 $\implies$ ```M'``` nunca imprime ```"Hola Mundo"```

4. Conectar los componentes
    * Ahora usamos nuestro supuesto ```DecisorHola``` para resolver la parada:

````python
def VerificadorDeParada(M, w):
    # Paso A: Fabricamos el programa M' descripto arriba
    M_prima = construir_M_prima(M, w)
    
    # Paso B: Le preguntamos a nuestro decisor hipotético
    resultado = DecisorHola(M_prima, "")
    
    # Paso C: Devolvemos la misma respuesta
    return resultado
````

* Si ```DecisorHola``` dice SÍ $\implies$ ```M'``` imprimió ```"Hola Mundo"``` $\implies$ ```M(w)``` se detuvo
* Si ```DecisorHola``` dice NO $\implies$ ```M'``` nunca imprimió ```"Hola Mundo"``` $\implies$ ```M(w)``` entró en bucle infinito

5. Conclusión
    * Habríamos creado ```VerificadorDeParada(M, w)```, un algoritmo perfecto para el problema de la parada. Pero Alan Turing demostró que tal algoritmo no puede existir
    * Por lo tanto, la suposición inicial es falsa: ```DecisorHola``` no puede existir y el problema $L_{\text{hola}}$ es indecidible
